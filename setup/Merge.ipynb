```python
import xarray as xr
```

```python
ds1 = xr.open_dataset('./../cpmip-prod/profiles.001.nc')
ds1
ds_out = ds1[['ta','thv','qt','ql','qr','u','v','w2r','skew','wthvr', 'rhof', 'presh']]
ds_out['qv'] = ds_out['qt'] - ds_out['ql']
ds_out['qv'].attrs['long_name'] = 'Specific humidity'
ds_out['qv'].attrs['units'] = 'kg/kg'
ds_out.to_netcdf('./../cpmip-prod/profiles-out.001.nc', 'w')
```
